# Part I — JOIN Foundations

## 1. INNER JOIN — Orders and Customers

Return the following for the first 20 orders:

- `order_id`
- `customer_id`
- `customer_city`
- `customer_state`
- `order_status`

Join `orders` to `customers` using the appropriate key.

**Concept:** `INNER JOIN`

````python
run_sql("""
    SELECT 
        o.order_id,
        o.customer_id,
        c.customer_city,
        c.customer_state,
        o.order_status
    FROM orders AS o
    INNER JOIN customers AS c
        ON o.customer_id = c.customer_id
    LIMIT 20;
""")
````

# Part II — GROUP BY, HAVING, WHERE, ORDER BY

## 5. Count Orders by Status

Return each `order_status` and the number of orders with that status.

Sort from the most common status to the least common status.

**Concepts:** `GROUP BY`, aggregate function, `ORDER BY`

## 6. Customers With More Than One Order

Find customers who placed more than one order.

Return:

- `customer_id`
- number of orders

Only return customers with more than one order.

**Concepts:** `GROUP BY`, `HAVING`

````python
run_sql("""
    SELECT 
        customer_id, 
        COUNT(order_id) AS number_of_orders
    FROM 
        orders 
    GROUP BY customer_id
    HAVING (number_of_orders > 1)
    LIMIT 10;
""")
````

````python
run_sql("""
    SELECT 
        customer_unique_id,
        COUNT(order_id) AS number_of_orders
    FROM 
        customers AS c
    LEFT JOIN 
        orders as o
        ON c.customer_id = o.customer_id
    GROUP BY customer_unique_id
    HAVING (number_of_orders > 5);
""")
````

## 7. WHERE vs HAVING

Find the number of delivered orders for each customer.

Only consider orders whose status is `delivered`.

Return customers with at least 2 delivered orders.

Return:

- `customer_id`
- delivered order count

**Concepts:** `WHERE` + `GROUP BY` + `HAVING`

````python
run_sql("""
    SELECT 
        c. customer_unique_id,
        COUNT(order_status) AS delivered_orders
    FROM 
        orders AS o
    RIGHT JOIN 
        customers AS c
        ON o.customer_id = c.customer_id
    WHERE 
        o.order_status = 'delivered'
    GROUP BY 
        c.customer_unique_id
    HAVING 
        (delivered_orders >= 2)
    LIMIT 10;
""")
````

## 8. Average Product Price by Category

Return each product category and its average product price.

Only include categories whose average product price is greater than 100.

Sort from highest average price to lowest.

**Concepts:** `GROUP BY`, `HAVING`, `ORDER BY`

````python
run_sql("""
    SELECT 
        p.product_category_name,
        AVG(price) AS avg_product_price
    FROM products AS p
    LEFT JOIN order_items AS i
        ON p.product_id = i.product_id
    GROUP BY p.product_category_name
    HAVING avg_product_price > 100
    ORDER BY avg_product_price DESC;
""")
````

````python
run_sql("""
    SELECT 
        p.product_category_name,
        ROUND(AVG(price), 3) AS avg_product_price
    FROM products AS p 
    LEFT JOIN order_items AS i
        ON p.product_id = i.product_id
    GROUP BY p.product_category_name
    HAVING avg_product_price > 100
    ORDER BY avg_product_price DESC;
""")
````

# Part III — Basic Subqueries

## 9a. Non-Correlated Subquery — Above-Average Product Price

Find products whose price is greater than the average price of **all products**.

Return:

- `product_id`
- `product_category_name`
- `price`

Sort from most expensive to least expensive.

**Concept:** non-correlated subquery

````python
run_sql("""
    SELECT
        o.product_id,
        o.price,
        p.product_category_name
    FROM order_items AS o
    INNER JOIN products AS p
        ON o.product_id = p.product_id
    WHERE o.price > (
        SELECT AVG(price)
        FROM order_items
        )
    ORDER BY o.price DESC
    LIMIT 10
""")
````

````python
run_sql("""
    SELECT 
        price,
        product_id
    FROM order_items
    WHERE price > 100
    ORDER BY price DESC
    LIMIT 10
""")
````

## 9b — Non-Correlated Subquery

Find order items whose price is below the average price of **all order items**.

Return:

- `product_id`
- `product_category_name`
- `price`

Sort from cheapest to most expensive.

**Concept:** non-correlated subquery

````python
run_sql("""
    SELECT 
        o.product_id,
        p.product_category_name,
        o.price
    FROM order_items AS o
    INNER JOIN products AS p
        ON o.product_id = p.product_id
    WHERE o.price < (
        SELECT AVG(price)
        FROM order_items
        )
    ORDER BY o.price
    LIMIT 10    
""")
````

## 10. Subquery with IN

Find all customers who placed at least one order.

Use a subquery with `IN`.

Repeat Question 10, but this time use `EXISTS` instead of `IN`.

Return the same customer information.

**Concept:** `EXISTS` + subquery

````python
run_sql("""
    SELECT 
        c.customer_id,
        c.customer_city,
        c. customer_state
    FROM customers AS c
    WHERE EXISTS(
        SELECT 1
        FROM orders AS o
        WHERE c.customer_id = o.customer_id)
    LIMIT 10
""")
````

## 21. Explain the Processing Order

Consider:

```sql
SELECT
    c.customer_state,
    COUNT(*) AS order_count
FROM customers AS c
JOIN orders AS o
    ON c.customer_id = o.customer_id
WHERE o.order_status = 'delivered'
GROUP BY c.customer_state
HAVING COUNT(*) >= 1000
ORDER BY order_count DESC
LIMIT 5;
```

Without executing it, explain what each stage is doing.

Use this conceptual order:

- FROM
- JOIN / ON
- WHERE
- GROUP BY
- HAVING
- SELECT
- ORDER BY
- LIMIT

Then explain why WHERE filters individual rows while HAVING filters groups.

**Concept:** logical query processing order

> Write your explanation in a markdown cell below or as comments.

````python
# Optional: run the query to verify your understanding
query = """
SELECT
    c.customer_state,
    COUNT(*) AS order_count
FROM customers AS c
JOIN orders AS o
    ON c.customer_id = o.customer_id
WHERE o.order_status = 'delivered'
GROUP BY c.customer_state
HAVING COUNT(*) >= 1000
ORDER BY order_count DESC
LIMIT 5;
"""

pd.read_sql_query(query, conn)
````
