```python
x = 3
y = 2
print(x + y)
```

```python
# Convert Input to Integer
x = int(input("What's x?"))
y = int(input("What's y?"))

z = int(x + y)

print(z)
```

```python
x = int(input("What's x?"))
y = int(input("What's y?"))

z = int(x) + int(y)

print(z)
```

```python
x = int(input("What's x"))
y = int(input("What's y"))

z = (x + y)

print(z)
```

```python
"""Introducing Float"""
x = float(input("What's x?"))
y = float(input("What's y?"))

z = (x + y)

print(z)
```

```python
z = round(x + y)
print(f"{z:,}")
```

```python
# Division
x = float(input("What's the value of x?"))
y = float(input("What's the value of y?"))

z = (x / y)
print(z)
```

```python
# Division
x = float(input("What's the value of x?"))
y = float(input("What's the value of y?"))

z = round(x / y, 2)
print(z)
```

```python
# Division without round off
x = float(input("What's the value of x?"))
y = float(input("What's the value of y?"))

z = (x / y) # or z = x / y
print(f"{z:.2f}")
```
