## Section 6: CTEs (Common Table Expressions)

**Question 42:** Write a query using a CTE to calculate the total revenue per user, then find users above the average total revenue.

```python
-- Your SQL query here
WITH user_revenue AS (
    SELECT o.user_id, SUM(o.grand_total) AS total_spent
    FROM orders o
    WHERE o.status = 'delivered'
    GROUP BY o.user_id
),
avg_revenue AS (
    SELECT AVG(total_spent) AS avg_spent FROM user_revenue
)
SELECT ur.user_id, ur.total_spent
FROM user_revenue ur, avg_revenue ar
WHERE ur.total_spent > ar.avg_spent
ORDER BY ur.total_spent DESC;
```

```python
-- Your SQL query here
SELECT 
    u.user_id,
    u.first_name,
    u.last_name,
    SUM(o.grand_total) AS total_spent,
    RANK() OVER (ORDER BY SUM(o.grand_total) DESC) AS spending_rank
FROM users u
JOIN orders o ON u.user_id = o.user_id
WHERE o.status = 'delivered'
GROUP BY u.user_id, u.first_name, u.last_name;
```

**Question 73:** Write a query to build a full order summary table including:
- Order ID
- Order Date
- Customer Name
- Customer Email
- Total Items
- Subtotal
- Discount Total
- Shipping Total
- Grand Total
- Payment Provider
- Shipping Carrier
- Tracking Number

```python
-- Your SQL query here
SELECT 
    o.order_id,
    o.ordered_at,
    u.first_name || ' ' || u.last_name AS customer_name,
    u.email,
    COUNT(oi.variant_id) AS total_items,
    o.subtotal,
    o.discount_total,
    o.shipping_total,
    o.grand_total,
    p.provider AS payment_provider,
    c.name AS shipping_carrier,
    s.tracking_number
FROM orders o
JOIN users u ON o.user_id = u.user_id
LEFT JOIN order_items oi ON o.order_id = oi.order_id
LEFT JOIN payments p ON o.order_id = p.order_id
LEFT JOIN shipments s ON o.order_id = s.order_id
LEFT JOIN carriers c ON s.carrier_id = c.carrier_id
GROUP BY o.order_id, o.ordered_at, u.first_name, u.last_name, u.email,
         o.subtotal, o.discount_total, o.shipping_total, o.grand_total,
         p.provider, c.name, s.tracking_number
ORDER BY o.ordered_at DESC
LIMIT 100;
```
