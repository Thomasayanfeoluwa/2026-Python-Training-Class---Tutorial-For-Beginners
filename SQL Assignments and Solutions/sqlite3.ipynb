```python
import sqlite3
import pandas as pd

# Connect to a SQLite database file
conn = sqlite3.connect("ecommerce.db")
```

```python
import os
print(os.getcwd())  # shows current working directory
```

```python
# Load CSV into pandas
df = pd.read_csv("../data/ecommerce_dataset.csv") 
```

```python
# Save it as a SQL table
df.to_sql("orders", conn, if_exists="replace", index=False)

print(f"✅ Done! Loaded {len(df):,} row into SQLite")
print(df.columns.tolist()) # This is to see all columns
```

```python
# Create a Helper Function

def run_sql(query):
    return pd.read_sql(query, conn)
```

```python
run_sql("SELECT * FROM orders;")
```

```python
run_sql("SELECT city FROM orders;")
```

```python
run_sql("SELECT order_id FROM orders;")
```

```python
run_sql("SELECT payment_method FROM orders;")
```

```python
run_sql("SELECT * FROM orders LIMIT 5;")
```

```python
run_sql("PRAGMA table_info(orders);")
```

```python
run_sql("""
    SELECT category, city, product, total_amount
    FROM orders
    WHERE city = 'Lagos'
    ORDER BY total_amount DESC
    LIMIT 20;
"""
)
```

```python
run_sql("""
    SELECT order_status, quantity, city, product
    FROM orders
    WHERE city = 'Ibadan'
    ORDER BY quantity DESC
    LIMIT 20;
""")
```

```python
run_sql("""
    SELECT order_id, customer_id, order_date, quantity, category, unit_price, payment_method 
    FROM orders
    WHERE payment_method = 'Bank Transfer'
    ORDER BY order_date DESC
    LIMIT 30
""")




# order_id	customer_id	order_date	product	category	
# quantity	unit_price	discount_percent	
# total_amount	city	payment_method	order_status
```

```python
run_sql("""
    SELECT category, quantity, payment_method
    FROM orders
    WHERE quantity = 4
    ORDER BY quantity DESC
    LIMIT 30
""")
```
