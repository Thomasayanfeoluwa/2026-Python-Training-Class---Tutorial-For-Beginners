### LIKE
- **`LIKE`**: Used for **text pattern matching** using wildcards (like `%` for multiple characters and `_` for a single character) to find specific words or letters.

```python
# Find all orders placed by customers whose city starts with the letter 'L', and order by customer_id descending
run_sql("""
    SELECT customer_id, city
    FROM orders
    WHERE city LIKE 'L%'
    ORDER BY customer_id DESC
    LIMIT 10;
""")
```

```python
# Retrieve all records where the order_status or description contains the text 'cell' anywhere inside it.
run_sql("""
    SELECT  order_date, product, order_status
    FROM orders
    WHERE order_status LIKE '%cell%'
    ORDER BY order_date
    LIMIT 20;
""")
```

```python
# Retrieve all records where the product or description contains the text 'Head' anywhere inside it.
run_sql("""
    SELECT order_date, customer_id, product
    FROM orders
    WHERE product LIKE '%Head%'
    ORDER BY order_date DESC
    LIMIT 20;
""")
```

### SQL Comparison & Logical Operators Cheat Sheet

- **`>` (Greater Than)**: Filters for values strictly higher than a threshold (e.g., `quantity > 5`).
- **`<` (Less Than)**: Filters for values strictly lower than a threshold (e.g., `unit_price < 100`).
- **`>=` (Greater Than or Equal To)**: Filters for values higher than or matching a threshold (e.g., `quantity >= 4`).
- **`<=` (Less Than or Equal To)**: Filters for values lower than or matching a threshold (e.g., `total_amount <= 500`).
- **`BETWEEN`**: Filters for a inclusive range between two values (e.g., `WHERE unit_price BETWEEN 100 AND 500`).
- **`AND`**: Logical operator requiring **all** conditions to be true simultaneously across different criteria/columns.
- **`OR`**: Logical operator requiring **at least one** condition to be true.

```python
"""
Challenge 1: The Multi-Condition Regional Filter
Write a query to find all orders where:
1. The city is either 'Lagos', 'Abuja', or 'Port Harcourt' (hint: think of a clean shortcut operator for this).
2. The quantity is greater than or equal to 3.
3. The order status is not 'Cancelled'.
Order the results by total_amount from highest to lowest, and limit the output to 15.
"""

run_sql("""
    SELECT city, quantity, order_status, total_amount
    FROM orders
    WHERE city IN ('Lagos', 'Abuja', 'Port Harcourt') AND quantity >= 3 AND order_status = 'Cancelled'
    ORDER BY total_amount DESC
    LIMIT 15;
""")
```

```python
"""
Challenge 2: Date Range & Financial Thresholds
Write a query to retrieve orders where:
1. The order_date falls within the second half of the year (between '2025-07-01' and '2025-12-31').
2. The total_amount is strictly greater than 500.00 OR the discount_percent is greater than 15.
Include only the order_id, order_date, total_amount, and discount_percent columns.
"""

run_sql("""
    SELECT order_id, order_date, total_amount, discount_percent
    FROM orders
    WHERE (order_date >= '2025-07-01' AND order_date <= '2025-12-31') AND (total_amount > 500.00 OR discount_percent > 15);
""")
```

```python
run_sql("""
    SELECT order_id, order_date, total_amount, discount_percent, quantity
    FROM orders
    WHERE order_date BETWEEN '2025-07-01' AND '2025-12-31' AND (total_amount > 500.00 OR discount_percent > 15)
    ORDER BY quantity DESC, total_amount DESC;
""")
```

```python
"""
Challenge 3: Text Pattern Matching (LIKE & Wildcards)
Write three distinct queries using LIKE:
1. Find all records where the product name starts with the letter 'S' and ends with any characters.
2. Find all records where the payment_method contains the word 'Transfer' anywhere inside it.
3. Find orders where the category has a word containing the exact pattern where the second character is 'e' (e.g., using _e%).
"""

run_sql("""
SELECT product, payment_method, category
FROM orders
WHERE product LIKE 'S%' AND payment_method LIKE '%Transfer' AND category LIKE '_e%';
""")
```

### SQL Aliases (`AS`)
- **`AS`**: Used to rename columns or tables temporarily in the output result for better readability or reporting (e.g., `SELECT COUNT(*) AS total_count`). 
- **Note**: The alias name appears as the column header in your final DataFrame or table output.

### SQL Aggregate Functions Reference

- **`COUNT()`**: Counts the total number of rows or non-null values in a column (e.g., `COUNT(*)` or `COUNT(unit_price)`).
- **`AVG()`**: Calculates the mathematical average (mean) for a numeric column, ignoring null values (e.g., `AVG(total_amount)`).
- **`MIN()`**: Finds the lowest/smallest value in a numeric, text, or date column (e.g., `MIN(order_date)` or `MIN(unit_price)`).
- **`MAX()`**: Finds the highest/largest value in a numeric, text, or date column (e.g., `MAX(total_amount)`).
- **`SUM()`**: Adds up all the values in a numeric column to give a grand total (e.g., `SUM(quantity)` or `SUM(total_amount)`).

### SQL Duplicate Removal (`DISTINCT`)

- **`DISTINCT`**: Used in a `SELECT` statement to filter out duplicate rows and return only **unique** values (e.g., `SELECT DISTINCT category FROM orders`).
- **Multi-Column Note**: When applied to multiple columns, it returns unique combinations of all those columns together.

```python
"""
Challenge 1: The Inventory & Sales Audit (SUM, MAX, MIN)
Write a single query that calculates the following metrics for the entire table:
1. The absolute maximum total_amount recorded in a single order, aliased as highest_sale.
2. The absolute minimum total_amount recorded, aliased as lowest_sale.
3. The grand sum of all quantities sold across all rows, aliased as total_units_sold.
"""

run_sql("""
    SELECT MAX(total_amount) AS highest_sale, 
    MIN(total_amount) AS lowest_sale, 
    SUM(quantity) AS total_units_sold
    FROM orders;
""")
```
