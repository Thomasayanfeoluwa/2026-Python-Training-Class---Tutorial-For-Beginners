```python
# Load CSV into pandas
df = pd.read_csv("data/ecommerce_dataset.csv")
```

```python
# Save it as a SQL table
df.to_sql("orders", conn, if_exists="replace", index=False)
print(f"✅ Done! Loaded {len(df):,} rows into SQLite")
print(f"Tables in SQLite {df.columns.to_list()}")
```

### IS NULL - IS NOT NULL TO CHECK FOR MISSING VALUES

- **`IS NULL` / `IS NOT NULL`**: Used to check for missing or populated values (presence/absence of data).


```python
run_sql("""
    SELECT order_date, order_id, city, quantity, unit_price
    FROM orders
    WHERE unit_price IS NULL;
""")
```

```python
run_sql("""
    SELECT order_date, order_id, city, quantity, unit_price
    FROM orders
    WHERE unit_price IS NOT NULL
    LIMIT 20;
""")
```

### LIKE
- **`LIKE`**: Used for **text pattern matching** using wildcards (like `%` for multiple characters and `_` for a single character) to find specific words or letters.

```python
# Find all orders placed by customers whose city starts with the letter 'L', and order by customer_id descending
run_sql("""
    SELECT customer_id, city
    FROM orders
    WHERE city LIKE 'L%'
    ORDER BY customer_id DESC
    LIMIT 10;
""")
```

```python
# Retrieve all records where the order_status or description contains the text 'cell' anywhere inside it.
run_sql("""
    SELECT  order_date, product, order_status
    FROM orders
    WHERE order_status LIKE '%cell%'
    ORDER BY order_date
    LIMIT 20;
""")
```

```python
# Retrieve all records where the product or description contains the text 'Head' anywhere inside it.
run_sql("""
    SELECT order_date, customer_id, product
    FROM orders
    WHERE product LIKE '%Head%'
    ORDER BY order_date DESC
    LIMIT 20;
""")
```

```python
# Find all orders where the city ends with 'dan' and oder got cancelled.
run_sql("""
    SELECT customer_id, product, city, order_status
    FROM orders
    WHERE city LIKE '%dan%' AND order_status = 'Cancelled';
""")
```

```python
# Select all rows where a text column contains a specific keyword (e.g., 'as' in a payment method or city name).
run_sql("""
    SELECT *
    FROM orders
    WHERE payment_method LIMIT '%as%' AND city LIMIT '%no'
""")
```

```python
# Find all records where a city name has exactly 5 letters (e.g., 'Abuja' or 'Kano ').
run_sql("""
    SELECT *
    FROM orders
    WHERE city LIMIT '_____';
""")
```

```python
# Find a product code or ID where the third character is always a specific letter or number (e.g., __X% or similar pattern).
run_sql("""
    SELECT customer_id, product
    FROM orders
    WHERE customer_id LIMIT '___3%' OR product LIMIT '___s%',
""")
```

```python
# Find customers whose city are exactly 4 letters long.
run_sql("""
    SELECT city
    FROM orders
    WHERE city LIMIT '___o';
""")
```

```python
# Retrieve orders where a specific 3-letter code appears at the beginning, followed by any characters.
run_sql("""
    SELECT order_id, product
    FROM orders
    WHERE product LIMIT 'Run____';
""")
```
